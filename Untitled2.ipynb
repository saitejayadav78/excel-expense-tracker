```python
#PROGRAM: Array joining using Stack Function
#Axis=0: (rows)   runs vertically downgrade across rows
#Axis=1: (columns)runs horizantally across columns
import numpy as np
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
print(" ---------AXIS-0-----------")
arr=np.stack((arr1,arr2),axis=0)
print(" Stacking Arrays : axis=0\n",arr)
print(" ----------AXIS-1-----------")
arr=np.stack((arr1,arr2),axis=1)
print(" Stacking Arrays : axis=1\n",arr)
#HStack
print(" ---------HStack-----------")
arr=np.hstack((arr1,arr2))
print(" HStack Arrays : \n",arr)
#VStack
print(" ---------VStack-----------")
arr=np.vstack((arr1,arr2))
print(" VStack Arrays : \n",arr)
#DStack
print(" ---------DStack-----------")
arr=np.dstack((arr1,arr2))
print(" DStack Arrays : \n",arr)
```

```python
#PROGRAM: NumPy Array Joining using concatenate
#Axis=0: (rows)   runs vertically downgrade across rows
#Axis=1: (columns)runs horizantally across columns

import numpy as np
#1-Dim Array join
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
arr=np.concatenate((arr1,arr2),axis=0)
print(" 1-Dim array Concatenate: ",arr)
print(" ---------------------------")
#2-Dim Array join
arr1=np.array([[1,2],
               [3,4]])
arr2=np.array([[5,6],
               [7,8]])
arr=np.concatenate((arr1,arr2),axis=0)
print(" 2-Dim array Concatenate: axis-0\n",arr)
print(" ---------------------------")
arr=np.concatenate((arr1,arr2),axis=1)
print(" 2-Dim array Concatenate: axis-1\n",arr)
```

```python
import numpy
import matplotlib.pyplot as plt
from scipy import stats

# x: Car Ages
# y: Car Speeds

x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]
slope, intercept, r, p, std_err=stats.linregress(x,y)

def myfun(x):
  return slope * x + intercept  # y=mx+c

mymodel=list(map(myfun,x))
plt.scatter(x,y)        # to plot dots
plt.plot(x,mymodel)     # to plot a line
plt.show()
```
